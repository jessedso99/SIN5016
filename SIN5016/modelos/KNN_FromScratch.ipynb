```python
# Base de dados
# 1: Spam, 0: Não Spam

emails = [
    "Free money now click here",
    "Call me back later please",
    "Win a free cruise vacation",
    "Are we meeting for lunch",
    "Click here to clain your free reward"
]
values = [1, 0, 1, 0, 1] 
labels = {0:'n_spam', 1:'spam'}
```

```python
from sklearn.feature_extraction.text import CountVectorizer

vectorizer = CountVectorizer()
X = vectorizer.fit_transform(emails)

database = list(zip(X.toarray(), values))
```

```python
def DistanciaEuclidiana(novo_email):
    distancias  = [] # distacnia do novo animal entre todos os elementos

    for index, email in enumerate(X.toarray()):
        #calc dist
        dist = sum((novo_email - email)**2)
        distancias.append(dist)

    dist_classe = dict(zip(distancias, values))
    dist_classe_sorted = sorted(dist_classe.items(), key=lambda item: item[0])

    return dist_classe_sorted

def K_NearestNeighbors(novo_email, alfa):
    Distancias_Classe = DistanciaEuclidiana(novo_email)

    # top (alfa) animais
    top_k_distancias = Distancias_Classe[:alfa]

    top_k_classes = []
    for i in top_k_distancias:
        top_k_classes.append(i[1])

    import statistics
    y = statistics.mode(top_k_classes)

    # atribuir nome da classe (dicionário = {0:'a', 1:'b', 2:'c'})
    classe_predita = labels.get(y)

    return classe_predita
```

```python
# Nova frase para prediçção e número de K vizinhos que queremos comparar

novo_email = ['Lets join in a meeting today']
alfa = 5 

texto_vectorizado = vectorizer.transform(novo_email).toarray()[0]
classe_predicao = K_NearestNeighbors(texto_vectorizado, alfa)

print(f"O e-mail foi classificado como: {classe_predicao}")
```
